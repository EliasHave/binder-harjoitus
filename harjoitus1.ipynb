```python
import pandas as pd

url = (
    "https://raw.githubusercontent.com/plotly/datasets/master/gapminder_with_codes.csv"
)
gapminder_data = pd.read_csv(url).query("year == 2007")

gapminder_data
```

```python
import matplotlib.pyplot as plt
import math

fig, ax = plt.subplots()

ax.scatter(x=gapminder_data["gdpPercap"], y=gapminder_data["lifeExp"], alpha=0.5)

ax.set_xlabel("GDP per capita (PPP dollars)")
ax.set_ylabel("Life expectancy (years)")
```

```python
fig, ax = plt.subplots()

ax.scatter(x=gapminder_data["gdpPercap"], y=gapminder_data["lifeExp"], alpha=0.25) #Alpha säätelee pisteiden läpinäkyvyyttä

ax.set_xscale("log")

ax.set_xlabel("GDP per capita (PPP dollars)")
ax.set_ylabel("Life expectancy (years)")
```

```python
fig, ax = plt.subplots()

ax.scatter(x=gapminder_data["gdpPercap"], y=gapminder_data["lifeExp"], alpha=0.25) #Alpha säätelee pisteiden läpinäkyvyyttä

ax.set_xscale("log")

ax.set_xlabel("GDP per capita (PPP dollars)", fontsize=16)
ax.set_ylabel("Life expectancy (years)", fontsize=16)

ax.tick_params(axis='both', labelsize=14, length=8, width=2)
```
